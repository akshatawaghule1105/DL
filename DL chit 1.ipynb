```python
import numpy as np
import tensorflow as tf
import torch

print("Libraries imported successfully!")
```

```python
a = tf.constant([[1.,2.],[3.,4.]])
b = tf.constant([[5.,6.],[7.,8.]])
print("Addition \n ",(a + b).numpy())
print("Multiplication \n", (a * b).numpy())
print("Matrix Multiplication \n", tf.matmul(a,b).numpy())
```

```python
a = torch.tensor([[1.,2.],[3.,4.]])
b = torch.tensor([[5.,6.],[7.,8.]])

print("Addition \n",(a + b).numpy())
print("Multiplication \n",(a * b).numpy())
print("Matrix Multiplication \n", torch.matmul(a,b).numpy())
```

```python
x = tf.Variable (3.0)
with tf.GradientTape() as tape  :
    y = x ** 2
    gradient = tape.gradient(y,x)
    print("TensorFlow Gradient : ", gradient.numpy())
```

```python
x = torch.tensor(3.0,requires_grad=True)
y = x ** 2
y.backward()
print("PyTorch Gradient :", x.grad.item())
```

```python
X = np.array([0,1,2,3,4], dtype= np.float32)
Y = 2 * X + 1
X = X.reshape(-1,1)
Y = Y.reshape(-1,1)
print("Input :",X.flatten())
print("Output :",Y.flatten())
```

```python
model_tf = tf.keras.Sequential ([
    tf.keras.Input(shape=(1,)),
    tf.keras.layers.Dense(1)
])
model_tf.compile(
    optimizer = tf.keras.optimizers.Adam(learning_rate=0.05),
    loss = "mse"
) 
model_tf.fit(X, Y, epochs=250, verbose=0)

prediction = model_tf.predict(
    np.array([[5.]],dtype=np.float32),verbose=0
)
print("Tensorflow Prediction :",prediction[0][0])
```

```python
X_torch = torch.tensor(X)
Y_torch = torch.tensor(Y)
model_torch = torch.nn.Linear(1 , 1)
optimizer = torch.optim.Adam(
   model_torch.parameters(),lr=0.05 
)
loss_fn = torch.nn.MSELoss()
for epoch in range(250):
    optimizer.zero_grad()

    prediction = model_torch(X_torch)
    loss = loss_fn(prediction,Y_torch)

    loss.backward()
    optimizer.step()
```

```python
with torch.no_grad():
    result = model_torch(torch.tensor([[5.0]]))
    print("Pytorch Prediction :",result.item())
```
